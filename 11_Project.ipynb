### Handling Missing Values

First, let's create a copy of the DataFrame to preserve the original.

Now, we will impute missing values:

*   **Numerical columns (`age`, `bmi`, `HbA1c_level`, `blood_glucose_level`):** Impute with the median.
*   **Binary numerical columns (`hypertension`, `heart_disease`):** Impute with the mode.
*   **Categorical columns (`gender`, `smoking_history`):** Impute with the mode.

### Handling Missing Values using `SimpleImputer`

Let's use `SimpleImputer` as requested. We'll impute numerical columns with the median and categorical columns with the most frequent value.

```python
from sklearn.impute import SimpleImputer

df_cleaned_imputed = df.copy()


numerical_cols_for_imputation = ['age', 'hypertension', 'heart_disease', 'bmi', 'HbA1c_level', 'blood_glucose_level', 'diabetes']
categorical_cols_for_imputation = cat_cols

median_imputer = SimpleImputer(strategy='median')
df_cleaned_imputed[numerical_cols_for_imputation] = median_imputer.fit_transform(df_cleaned_imputed[numerical_cols_for_imputation])

mode_imputer = SimpleImputer(strategy='most_frequent')
df_cleaned_imputed[categorical_cols_for_imputation] = mode_imputer.fit_transform(df_cleaned_imputed[categorical_cols_for_imputation])

print("Missing values after SimpleImputer:")
display(df_cleaned_imputed.isnull().sum())
```

###Implementing Neural Network

```python
import torch
from torch import nn

# Make device agnostic code
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
X_train_tensor = torch.from_numpy(X_train_scaled).type(torch.float).to(device)
X_test_tensor = torch.from_numpy(x_test_scaled).type(torch.float).to(device)

y_train_tensor = torch.from_numpy(y_train.values).type(torch.float).to(device)
y_test_tensor = torch.from_numpy(y_test.values).type(torch.float).to(device)
y_train_tensor[:5]
```

```python
class DiabetesModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer_1 = nn.Linear(in_features=9, out_features=128)
        self.layer_2 = nn.Linear(in_features=128, out_features=1)
        self.relu = nn.ReLU()

    def forward(self, x):
      # Intersperse the ReLU activation function between layers
       return self.layer_2(self.relu(self.layer_1(x)))

model_1 = DiabetesModel().to(device)
print(model_1)
```

```python
model_1.state_dict()
```

###Setting up Loss function ==> Binary Cross Entropy
### Using SGD for Optimizer

```python
loss_fn = nn.BCEWithLogitsLoss()
optimizer = torch.optim.SGD(model_1.parameters(), lr=0.1)
```

```python
# Calculate accuracy (a classification metric)
def accuracy_fn(y_true, y_pred):
    correct = torch.eq(y_true, y_pred).sum().item()
    acc = (correct / len(y_pred)) * 100
    return acc
```

```python
# Fit the model
torch.manual_seed(42)
epochs = 1000

train_loss_values = []
test_loss_values = []
epoch_count = []

for epoch in range(epochs):
    # 1. Forward pass
    model_1.train()
    y_logits = model_1(X_train_tensor).squeeze()
    y_pred = torch.round(torch.sigmoid(y_logits)) # logits -> prediction probabilities -> prediction labels

    # 2. Calculate loss and accuracy
    loss = loss_fn(y_logits, y_train_tensor) # BCEWithLogitsLoss calculates loss using logits
    acc = accuracy_fn(y_true=y_train_tensor,
                      y_pred=y_pred)

    # 3. Optimizer zero grad
    optimizer.zero_grad()

    # 4. Loss backward
    loss.backward()

    # 5. Optimizer step
    optimizer.step()

    ### Testing
    model_1.eval()
    with torch.inference_mode():
      # 1. Forward pass
      test_logits = model_1(X_test_tensor).squeeze()
      test_pred = torch.round(torch.sigmoid(test_logits)) # logits -> prediction probabilities -> prediction labels
      # 2. Calculate loss and accuracy
      test_loss = loss_fn(test_logits, y_test_tensor)
      test_acc = accuracy_fn(y_true=y_test_tensor,
                             y_pred=test_pred)

    # Print out what's happening
    if epoch % 100 == 0:
        epoch_count.append(epoch)
        train_loss_values.append(loss.detach().cpu().numpy())
        test_loss_values.append(test_loss.detach().cpu().numpy())
        print(f"Epoch: {epoch} | Loss: {loss:.5f}, Accuracy: {acc:.2f}% | Test Loss: {test_loss:.5f}, Test Accuracy: {test_acc:.2f}%")
```

```python
plt.plot(epoch_count, train_loss_values, label="Train loss")
plt.plot(epoch_count, test_loss_values, label="Test loss")
plt.title("Training and test loss curves")
plt.ylabel("Loss")
plt.xlabel("Epochs")
plt.legend();
```

```python
# Make predictions
model_1.eval()
with torch.inference_mode():
    y_preds = torch.round(torch.sigmoid(model_1(X_test_tensor))).squeeze()
```

### Unsupervised Learning: K-Means Clustering

```python
from sklearn.cluster import KMeans
def implement_KMC(df_temp,cluster_num):
  km = KMeans(n_clusters=cluster_num)
  y_predicted = km.fit_predict(df_temp)
  return km,y_predicted
```

```python
cluster = 4
# Scale the entire dataset X using the scaler fitted on X_train.
# Convert to DataFrame to maintain column names and index for consistency with original X.
X_scaled_for_kmeans = pd.DataFrame(scaler.transform(X), columns=X.columns, index=X.index)

km, y_predicted = implement_KMC(X_scaled_for_kmeans, cluster)
df_cleaned_imputed["Group_cluster"] = y_predicted
df_cleaned_imputed.head(10)
```

```python
from sklearn.decomposition import PCA
X_for_pca_scaled = X_scaled_for_kmeans.copy()
# Initialize PCA with 3 components for visualization
pca = PCA(n_components=4)
X_pca_components = pca.fit_transform(X_for_pca_scaled)
pca_df_new = pd.DataFrame(data=X_pca_components, columns=['PC1', 'PC2', 'PC3', 'PC4'])
pca_df_new['Group_cluster'] = df_cleaned_imputed['Group_cluster'].reset_index(drop=True)
print("New pca_df for plotting with 3 clusters, based on scaled data:")
display(pca_df_new.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x='PC1', y='PC2', hue='Group_cluster', data=pca_df_new, palette='viridis', legend='full')
centroids_original_space = km.cluster_centers_
centroids_pca_space = pca.transform(centroids_original_space)

# Plot the centroids
plt.scatter(
    centroids_pca_space[:, 0],
    centroids_pca_space[:, 1],
    marker='X',
    s=200,
    color='red',
    edgecolor='black',
    label='Centroids'
)

plt.title('K-Means Clustering with Centroids (PCA Reduced on Scaled Data)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.legend()
plt.show()
```

### Elbow Method to find Optimal Number of Clusters



```python
wcss = []
for i in range(1, 11):
    km_elbow = KMeans(n_clusters=i, random_state=42, n_init='auto')
    km_elbow.fit(X_scaled_for_kmeans)
    wcss.append(km_elbow.inertia_)

# Plot the Elbow Method graph
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```
